```python
import pandas as pd
```

```python
file_name = "Day11_Messy_Company_Employee_Dataset.csv"
employee_data = pd.read_csv(file_name)
print("Dataset loaded successfully.")
```

```python
print("========== COMPANY EMPLOYEE DATASET ==========")
display(employee_data.head())
```

```python
print("========== DATASET SHAPE ==========")
print(f"Number of rows: {employee_data.shape[0]}")
print(f"Number of columns: {employee_data.shape[1]}")
print(f"Dataset shape: {employee_data.shape}")
```

```python
print("========== COLUMN NAMES ==========")
for column_name in employee_data.columns:
  print("-", column_name)
```

```python
print("========== DATA TYPES ==========")
print(employee_data.dtypes)
```

```python
print("========== DATASET INFROMATION ==========")
employee_data.info()
```

```python
print("========== DESCRIPTIVE STATISTICS ==========")
display(employee_data.describe())
```

```python
print("========== MISSING VALUES ==========")
missing_values = employee_data.isnull().sum()
display(missing_values)
```

```python
print("========== MISSING VALUE PERCENTAGE ==========")
missing_percentage = (
    employee_data.isnull().mean() * 100
).round(2)

missing_summary = pd.DataFrame(
    {
        "Missing_Count": missing_values,
        "Missing_Percentage": missing_percentage
    }
)

display(missing_summary)
```

```python
print("========== DUPLICATE ANALYSIS ==========")
duplicate_count = employee_data.duplicated().sum()
print(
    f"Number of duplicate rows: "
    f"{duplicate_count}"
)
```

```python
duplicate_records = employee_data[
    employee_data.duplicated(keep=False)
]

print("========== DUPLICATE RECORDS ==========")
display(duplicate_records)
```

```python
numeric_columns = [
    "Age",
    "Annual_Salary",
    "Experience_Years",
    "Performance_Score"
]

for column_name in numeric_columns:
  median_value = employee_data[column_name].median()

  employee_data[column_name] = (
      employee_data[column_name].fillna(
          median_value
      )
  )

print("Numeric missing values filled using median.")
```

```python
categorical_columns = [
    "Department",
    "Gender",
    "City",
    "Work_Mode"
]

for column_name in categorical_columns:
  mode_value = employee_data[column_name].mode()[0]

  employee_data[column_name] = (
      employee_data[column_name].fillna(
          mode_value
      )
  )

print(
    "Categorical missing values filled using mode."
)
```

```python
employee_data["Joining_Date"] = pd.to_datetime(
    employee_data["Joining_Date"],
    errors="coerce"
)

print("Joining_Date converted to datetime.")
print(
    employee_data["Joining_Date"].dtype
)
```

```python
missing_dates = employee_data[
    "Joining_Date"
].isnull().sum()
print(
    f"Missing or invalid joining dates: "
    f"{missing_dates}"
)
```

```python
rows_before_duplicates = len(employee_data)
employee_data = employee_data.drop_duplicates()
rows_after_duplicates = len(employee_data)
duplicates_removed = (
    rows_before_duplicates - rows_after_duplicates
)
print(
    f"Duplicate rows removed: "
    f"{duplicates_removed}"
)
print(
    f"Rows before removing duplicates: "
    f"{rows_before_duplicates}"
)
print(
    f"Rows after removing duplicates: "
    f"{rows_after_duplicates}"
)
```

```python
print("========== MISSING VALUES AFTER CLEANING ==========")
remaining_missing = employee_data.isnull().sum()
display(remaining_missing)
```

```python
print("========== DUPLICATES AFTER CLEANING ==========")
remaining_duplicates = (
    employee_data.duplicated().sum()
)
print(
    f"Remianing duplicate rows: "
    f"{remaining_duplicates}"
)
```

```python
print("========== DATA TYPES AFTER CLEANING ==========")
print(employee_data.dtypes)
```

```python
forward_fill_demo = employee_data[
    ["Employee_ID", "Joining_Date"]
].copy()
forward_fill_demo["Joining_Date"] = (
    forward_fill_demo["Joining_Date"].ffill()
)
print(
    "Forward-fill demonstration completed."
)
display(forward_fill_demo.head())
```

```python
print("========== BEFORE VS AFTER CLEANING ==========")
print(
    f"Original number of rows: "
    f"{rows_before_duplicates}"
)
print(
    f"Final mumber of rows: "
    f"{rows_after_duplicates}"
)
print(
    f"Duplicate rows removed: "
    f"{duplicates_removed}"
)
print(
    f"Remianing duplicate rows: "
    f"{employee_data.duplicated().sum()}"
)
print(
    f"Remaining missing values: "
    f"{employee_data.isnull().sum().sum()}"
)
```

```python
print("========== CLEANED DATASET STATISTICS ==========")
display(employee_data.describe())
```

```python
print("========== CLEANED EMPLOYEE DATASET ==========")
display(employee_data.head(10))
```

```python
print("========== FINAL DATASET INFORMATION ==========")
employee_data.info()
```

```python
output_file = "Day11_Cleaned_Company_Employee_Dataset.csv"
employee_data.to_csv(
    output_file,
    index=False
)
print(
    f"Cleaned dataset exported successfully as "
    f"{output_file}"
)
```
